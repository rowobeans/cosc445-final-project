```python
import pandas as pd
import glob
import os

csvFiles = sorted(glob.glob("data/ao3-*.csv"))
dataframes = []
for file in csvFiles:
    tempDF = pd.read_csv(file)
    dataframes.append(tempDF)

df = pd.concat(dataframes, ignore_index=True)
print("Number of CSV files:", len(csvFiles))
print("Total number of issues:", len(df))
print("Number of columns:", len(df.columns))
df.head()
```

```python
for column in df.columns:
    print(column)
```

```python
df["Created"] = pd.to_datetime(df["Created"].astype(str).str.strip(), format="%d/%b/%y %I:%M %p", errors="coerce")
```

```python
print("Valid dates:", df["Created"].notna().sum())
print("Invalid dates:", df["Created"].isna().sum())
print("Earliest date:", df["Created"].min())
print("Latest date:", df["Created"].max())
```

```python
df["Created Month"] = df["Created"].dt.to_period("M")
monthlyI = df.groupby("Created Month").size()
print("Number of months:", len(monthlyI))
print("\nFirst 12 months:")
print(monthlyI.head(12))
```

```python
import matplotlib.pyplot as plt

monthlyI.plot(figsize=(12, 5), title="AO3 Issues Created Per Month", xlabel="Month", ylabel="Number Of Issues")
plt.tight_layout()
plt.show()
```

```python
print("Issues priority:")
print(df["Priority"].value_counts(dropna=False))
print("\nIssues Type:")
print(df["Issue Type"].value_counts(dropna=False))
```

```python
df["Issue Type"].value_counts().plot(kind="bar", figsize=(10, 5), title="AO3 Issues Counts", xlabel="Issue Type", ylabel="Number Of Issues")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
groupR = []

for groupN, columns in groups.items():
    eColumns = []
    for column in columns:
        if column in df.columns:
            eColumns.append(column)
    groupD = df[eColumns]
    issuesU = groupD.notna().any(axis=1).sum()
    Tvalues = groupD.notna().sum().sum()
    columnsU = groupD.notna().any().sum()
    groupR.append({"Group": groupN, "Total Columns": len(eColumns), "Columns Used": columnsU, "Total Values": Tvalues, "Issues Using Group": issuesU, "Percent Of Issues": round(issuesU / len(df) * 100, 2)})

Gsummary = pd.DataFrame(groupR)
Gsummary
```

```python
Gsummary.plot(x="Group", y="Percent Of Issues", kind="bar", figsize=(12, 6), legend=False, title="Percentage Of AO3 Issues Using In Each Group")
plt.xlabel("Group")
plt.ylabel("Percentage Of Issues")
plt.xticks(rotation=45, ha="right")
plt.ylim(0, 100)
plt.tight_layout()
plt.show()
```

```python
groupCU = {}
pd.set_option("display.max_rows", None)
pd.set_option("display.max_columns", None)
pd.set_option("display.width", None)

for groupN, columns in groups.items():
    Ecolumns = []
    for column in columns:
        if column in df.columns:
            Ecolumns.append(column)

    usage = pd.DataFrame({"Filled": df[Ecolumns].notna().sum(), "Empty": df[Ecolumns].isna().sum()})
    usage["Percent Filled"] = (usage["Filled"] / len(df) * 100)
    usage = usage.sort_values("Filled", ascending=False)
    groupCU[groupN] = usage
    print("\n", groupN)
    display(usage)
```
